### Blue Dashed Line with Circle Markers

This plot demonstrates how to customize a line plot using a format string: `'bo--'`.

*   **`'b'`**: Sets the line color to blue.
*   **`'o'`**: Specifies that circle markers should be used at each data point.
*   **`'--'`**: Creates a dashed line connecting the data points.
*   **`linewidth=2`**: Adjusts the thickness of the line.

```python
import numpy as np

# Generate some random data for a scatter plot
n_points = 50
np.random.seed(42) # for reproducibility
scatter_x = np.random.rand(n_points) * 10
scatter_y = np.random.rand(n_points) * 20
colors = np.random.rand(n_points)
sizes = np.random.rand(n_points) * 100 + 20 # sizes from 20 to 120

plt.figure(figsize=(8, 6))
plt.scatter(scatter_x, scatter_y, c=colors, s=sizes, alpha=0.7, cmap='viridis')
plt.colorbar(label='Color intensity')
plt.xlabel('Random X-values')
plt.ylabel('Random Y-values')
plt.title('Random Scatter Plot with Varied Color and Size')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Random Scatter Plot with Varied Color and Size

This scatter plot visualizes the relationship between two random variables (`scatter_x` and `scatter_y`).

*   **`np.random.rand(n_points)`**: Generates arrays of random numbers, which are used for `x` values, `y` values, `color` values, and `size` values.
*   **`plt.scatter(...)`**: Creates the scatter plot.
    *   **`c=colors`**: Assigns a color to each point based on a third random variable (`colors`).
    *   **`s=sizes`**: Sets the size of each point based on another random variable (`sizes`).
    *   **`alpha=0.7`**: Makes the points semi-transparent, which helps in visualizing overlapping points.
    *   **`cmap='viridis'`**: Specifies the colormap to use for `c` values. 'viridis' is a perceptually uniform colormap.
*   **`plt.colorbar(label='Color intensity')`**: Adds a color bar to the plot, indicating the mapping of `colors` to the colormap.
*   **`plt.xlabel`, `plt.ylabel`, `plt.title`**: Set the labels for the axes and the title of the plot for clarity.
*   **`plt.grid(True, linestyle=':', alpha=0.6)`**: Adds a subtle grid to the background to help with reading values.

```python
categories = ['Category A', 'Category B', 'Category C', 'Category D']
values = [23, 45, 56, 12]

plt.figure(figsize=(7, 5))
plt.bar(categories, values, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Simple Bar Graph')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Simple Bar Graph

This bar graph is used to display and compare the quantities of different categories.

*   **`categories`**: A list of strings representing the categorical labels for the x-axis.
*   **`values`**: A list of numerical data corresponding to each category, determining the height of the bars.
*   **`plt.bar(categories, values, ...)`**: Creates the bar chart.
    *   **`color=['skyblue', 'lightcoral', 'lightgreen', 'gold']`**: Sets specific colors for each bar, making them visually distinct.
*   **`plt.xlabel`, `plt.ylabel`, `plt.title`**: Label the x-axis, y-axis, and set the title of the plot for clarity.
*   **`plt.grid(axis='y', linestyle='--', alpha=0.7)`**: Adds a horizontal grid to make it easier to read the values for each bar.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate a grid of X and Y values, same as previous plots
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)

# Define the Z-axis function for the surface
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a new figure and an axes object for 3D plotting
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Create the 3D surface plot (the 'heatframe' part)
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.8)

# Create the 3D wireframe plot on the same axes
# You can adjust 'rstride' and 'cstride' for density, and 'color' for visibility
ax.plot_wireframe(X, Y, Z, rstride=10, cstride=10, color='black', alpha=0.3)

# Set labels for the axes
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

# Set the title of the plot
ax.set_title('3D Wireframe with Surface (Heatmap)')

# Add a color bar to indicate Z-axis values for the surface
fig.colorbar(surf, shrink=0.5, aspect=5, label='Z-value')

# Display the plot
plt.show()
```
