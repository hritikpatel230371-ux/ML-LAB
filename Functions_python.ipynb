## Basic Python Functions: Examples

### 1. Functions with no arguments and no return value

```python
# Function 1: Simple greeting
def say_hello():
  print("Hello, world!")

say_hello()
```

```python
# Function 2: Display a message
def show_message():
  print("This is a simple message.")

show_message()
```

### 2. Functions with arguments

```python
# Function 3: Greet a specific name
def greet_user(name):
  print(f"Hello, {name}!")

greet_user("Alice")
```

```python
# Function 4: Add two numbers and print the result
def add_and_print(a, b):
  result = a + b
  print(f"The sum is: {result}")

add_and_print(10, 5)
```

### 3. Functions with default arguments

```python
# Function 5: Greet with a default name
def greet_default(name="Guest"):
  print(f"Welcome, {name}!")

greet_default()
greet_default("Bob")
```

```python
# Function 6: Power calculation with default exponent
def calculate_power(base, exponent=2):
  print(f"{base} raised to the power of {exponent} is: {base**exponent}")

calculate_power(5)
calculate_power(2, 3)
```

### 4. Functions with `*args` (arbitrary positional arguments)

```python
# Function 7: Sum any number of arguments
def sum_all(*numbers):
  total = 0
  for num in numbers:
    total += num
  print(f"Sum of numbers: {total}")

sum_all(1, 2, 3)
sum_all(10, 20, 30, 40)
```

### 5. Functions with `**kwargs` (arbitrary keyword arguments)

```python
# Function 8: Print user information
def print_info(**kwargs):
  for key, value in kwargs.items():
    print(f"{key.replace('_', ' ').title()}: {value}")

print_info(name="Charlie", age=30, city="New York")
```

### 6. Functions with return values

```python
# Function 9: Return the product of two numbers
def multiply(x, y):
  return x * y

product = multiply(7, 8)
print(f"Product: {product}")
```

```python
# Function 10: Return a capitalized string
def capitalize_string(text):
  return text.capitalize()

caps_text = capitalize_string("python programming")
print(caps_text)
```

### 7. Functions with conditional logic

```python
# Function 11: Check if a number is even or odd
def check_even_odd(number):
  if number % 2 == 0:
    print(f"{number} is even.")
  else:
    print(f"{number} is odd.")

check_even_odd(4)
check_even_odd(7)
```

```python
# Function 12: Determine the maximum of two numbers
def get_max(a, b):
  if a > b:
    return a
  else:
    return b

max_val = get_max(15, 8)
print(f"Maximum value: {max_val}")
```

### 8. Functions with loops

```python
# Function 13: Print numbers from 1 to n
def count_up(n):
  for i in range(1, n + 1):
    print(i)

count_up(3)
```

```python
# Function 14: Iterate through a list and print items
def print_list_items(items):
  for item in items:
    print(f"Item: {item}")

my_list = ["apple", "banana", "cherry"]
print_list_items(my_list)
```

### 9. Functions for basic math operations

```python
# Function 15: Subtract two numbers
def subtract(x, y):
  return x - y

print(f"Subtraction: {subtract(20, 7)}")
```

```python
# Function 16: Divide two numbers
def divide(x, y):
  if y == 0:
    return "Error: Division by zero"
  return x / y

print(f"Division: {divide(10, 2)}")
print(f"Division by zero: {divide(5, 0)}")
```

```python
# Function 17: Calculate remainder (modulo)
def get_remainder(x, y):
  return x % y

print(f"Remainder: {get_remainder(10, 3)}")
```

### 10. Functions for string manipulation

```python
# Function 18: Convert string to uppercase
def to_uppercase(text):
  return text.upper()

print(f"Uppercase: {to_uppercase('hello python')}")
```

```python
# Function 19: Concatenate two strings
def concatenate_strings(str1, str2):
  return str1 + " " + str2

print(f"Concatenation: {concatenate_strings('Good', 'Morning')}")
```

```python
# Function 20: Get the length of a string
def get_string_length(text):
  return len(text)

print(f"String length: {get_string_length('programming')}")
```

### 11. Functions for list manipulation

```python
# Function 21: Append an item to a list
def add_item_to_list(my_list, item):
  my_list.append(item)
  print(f"Updated list: {my_list}")

my_numbers = [1, 2, 3]
add_item_to_list(my_numbers, 4)
```

```python
# Function 22: Remove an item from a list
def remove_item_from_list(my_list, item):
  if item in my_list:
    my_list.remove(item)
    print(f"Updated list: {my_list}")
  else:
    print(f"{item} not found in list.")

my_colors = ["red", "green", "blue"]
remove_item_from_list(my_colors, "green")
remove_item_from_list(my_colors, "yellow")
```

```python
# Function 23: Get the length of a list
def get_list_length(my_list):
  return len(my_list)

animals = ["cat", "dog", "bird"]
print(f"List length: {get_list_length(animals)}")
```

### 12. More complex function examples

```python
# Function 24: Factorial (recursive example)
def factorial(n):
  if n == 0:
    return 1
  else:
    return n * factorial(n-1)

print(f"Factorial of 5: {factorial(5)}")
```

```python
# Function 25: Check for prime number
def is_prime(num):
  if num <= 1:
    return False
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      return False
  return True

print(f"Is 7 prime? {is_prime(7)}")
print(f"Is 10 prime? {is_prime(10)}")
```

### 13. Functions for collection manipulation (lists, tuples, dictionaries)

```python
# Function 26: Find the smallest number in a list
def find_min(numbers_list):
  if not numbers_list:
    return None
  return min(numbers_list)

my_numbers = [5, 2, 9, 1, 7]
print(f"Smallest number: {find_min(my_numbers)}")
```

```python
# Function 27: Find the largest number in a list
def find_max(numbers_list):
  if not numbers_list:
    return None
  return max(numbers_list)

my_numbers = [5, 2, 9, 1, 7]
print(f"Largest number: {find_max(my_numbers)}")
```

```python
# Function 28: Count occurrences of an item in a list
def count_item(my_list, item):
  return my_list.count(item)

my_items = ['a', 'b', 'a', 'c', 'a']
print(f"'a' appears {count_item(my_items, 'a')} times.")
```

```python
# Function 29: Check if a key exists in a dictionary
def check_key_in_dict(my_dict, key):
  return key in my_dict

my_dict = {'name': 'John', 'age': 30}
print(f"'name' in dict: {check_key_in_dict(my_dict, 'name')}")
print(f"'city' in dict: {check_key_in_dict(my_dict, 'city')}")
```

```python
# Function 30: Get dictionary keys as a list
def get_dict_keys(my_dict):
  return list(my_dict.keys())

my_dict = {'name': 'John', 'age': 30}
print(f"Dictionary keys: {get_dict_keys(my_dict)}")
```

### 14. Functions for basic numerical operations

```python
# Function 31: Calculate the average of a list of numbers
def calculate_average(numbers):
  if not numbers:
    return 0
  return sum(numbers) / len(numbers)

scores = [85, 90, 78, 92]
print(f"Average score: {calculate_average(scores)}")
```

```python
# Function 32: Convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
  return (celsius * 9/5) + 32

print(f"25 Celsius is {celsius_to_fahrenheit(25)} Fahrenheit.")
```

```python
# Function 33: Convert Fahrenheit to Celsius
def fahrenheit_to_celsius(fahrenheit):
  return (fahrenheit - 32) * 5/9

print(f"77 Fahrenheit is {fahrenheit_to_celsius(77)} Celsius.")
```

### 15. Functions for string and text processing

```python
# Function 34: Reverse a string
def reverse_string(s):
  return s[::-1]

print(f"Reversed string: {reverse_string('hello')}")
```

```python
# Function 35: Check if a string is a palindrome
def is_palindrome(s):
  cleaned_s = s.replace(' ', '').lower()
  return cleaned_s == cleaned_s[::-1]

print(f"'madam' is palindrome: {is_palindrome('madam')}")
print(f"'racecar' is palindrome: {is_palindrome('Racecar')}")
print(f"'hello' is palindrome: {is_palindrome('hello')}")
```

```python
# Function 36: Count words in a string
def count_words(text):
  words = text.split()
  return len(words)

my_sentence = "This is a sample sentence."
print(f"Word count: {count_words(my_sentence)}")
```

```python
# Function 37: Replace a substring
def replace_substring(original_string, old_sub, new_sub):
  return original_string.replace(old_sub, new_sub)

text = "I love programming."
print(f"After replacement: {replace_substring(text, 'love', 'enjoy')}")
```

### 16. Utility Functions

```python
# Function 38: Generate a range of numbers
def generate_range(start, end):
  return list(range(start, end + 1))

print(f"Generated range: {generate_range(1, 5)}")
```

```python
# Function 39: Print a formatted message with current time
import datetime
def log_message(message):
  current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
  print(f"[{current_time}] {message}")

log_message("Application started.")
```

```python
# Function 40: Check if a number is positive, negative, or zero
def check_number_sign(num):
  if num > 0:
    return "Positive"
  elif num < 0:
    return "Negative"
  else:
    return "Zero"

print(f"Sign of 5: {check_number_sign(5)}")
print(f"Sign of -3: {check_number_sign(-3)}")
print(f"Sign of 0: {check_number_sign(0)}")
```
